Theory:
A neuron computes:z=w1x1+w2x2+w3x3+.......+wnxn+b where x is inputs and w=weights and b=bias and z=output

```python
import numpy as np
```

```python
#Inputs:
X=np.array([2,3])
#weights:
W=np.array([0.5,0.8])
#Bias=1
b=1
#weighted sum
z=np.dot(X,W)+b #Here np.dot() becomes dot product
print("Output:",z)
```

```python
#Part-2 Multiple Neurons:
# 2inputs and 3 neurons:
#input:
X=np.array([2,3])
#weights for 3 neurons:
W=np.array([
    [0.5,0.8],
    [0.2,0.4],
    [0.9,0.1]
])
#biases:
b=np.array([1,1,1])
#Output:
z=np.dot(W,X)+b
print("Outputs:",z)
```

```python
#Part-3 Multiple Input samples(Batches)
#Real training never uses one sample:
#It uses batches:
#COde is:
#batch of inputs:
X=np.array([
    [2,3],
    [1,5],
    [4,2]
])
#weights:
W=np.array([
    [0.5,0.8],[0.2,0.4],[0.9,0.1]
])
#biases:
b=np.array([1,1,1])
#Forward pass:
z=np.dot(X,W.T)+b
print(z)
```

PART-4 BUILDING A LAYER:

```python
class DenseLayer:
  def __init__(self,input_size,output_size):
    #Random weights:
    self.weights=np.random.randn(output_size,input_size)
    #biases: intilaised to zero:
    self.biases=np.zeros((1,output_size))
  def forward(self,X):
    #Store Input:
    self.input=X
    #Layer output:
    self.output=np.dot(X,self.weights.T)+self.biases
    return self.output
```

```python
X=np.array([
    [2,3],
    [1,5],
    [4,2]
])
#Create a layer:
layer=DenseLayer(input_size=2,output_size=3)
#Forward pass
output=layer.forward(X)
print(output)
```
